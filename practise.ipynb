```python
#create a new column salary_lakhs by converting salary_inr into lakhs.

from pyspark.sql.functions import round

df_pyspark.withColumn("salary_lakhs", round(
    col("salary_inr")/100000,
    5
    )
    ).show()
```

```python
#experience_category:

from pyspark.sql.functions import col, when

df_pyspark.withColumn(
    "ExperinceCategory",
                      when(col("years_experience") < 5 , "Junior")
                      .when(col("years_experience") < 10, "Mid")
                      .when(col("years_experience") <= 15, "High")
                      .otherwise("Expert")
).show()
```

```python
'''create salary_category:
salary < ₹1,000,000 → "low"
₹1,000,000–₹2,000,000 → "medium"
₹2,000,000 → "high"'''




df_pyspark.withColumn("SalCategory",
                      when(col("salary_inr")<1000000 , "Low")
                      .when(col("salary_inr") < 200000 , "Mid")
                      .when(col("salary_inr") < 300000 , "High")
                      .otherwise("Exceptional")
                      ).show()
```

```python
#create a total_compensation column using salary + monthly bonus points.


df_pyspark.withColumn("TotalComp",
                      col("salary_inr") + col("Monthly_Bonus_points")
                      ).show()
```

```python
#create projects_per_year using projects_completed / years_experience.


df_pyspark.withColumn("ProjPerYear",round(
                      col("projects_completed") / col("years_experience"),2)
                      ).show()
```

```python
'''create a column performance_category:
performance score < 2 → "poor"
2–3 → "average"
3 → "good"
'''


df_pyspark.withColumn("PerformanceCat",
                      when(col("performance_score") < 2 , "Poor")
                      .when(col("performance_score") < 3 , "Avg")
                      .when(col("performance_score") < 4 , "Good")
                      .otherwise("Excellent")
                      ).show()
```

```python
#create experience_salary_ratio using salary and years of experience.


df_pyspark.withColumn("ExpSalRatio",
                      round(col("salary_inr") / col("years_experience"),2)
).show()

```

```python
#create employee_tenure using join_date and today's date.

from pyspark.sql.functions import col, datediff, current_date

df_pyspark.withColumn("EmpTenure",
                      datediff(current_date(),"join_date")
                      ).show()
```

```python
#create review_delay_days showing the number of days between join_date and last_review_date.


df_pyspark.withColumn("ReviewDelayDays",
                      datediff("last_review_date","join_date")
).show()
```

```python
#create high_performer where employees with performance score ≥ 4 are "yes", otherwise "no".



df_pyspark.withColumn("HighPerformer",
                      when(col("performance_score") >=4, "Yes")
                      .otherwise("No")
                      ).show()



```
